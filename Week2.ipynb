# For loop

A for loop repeats a block of code for each item in a sequence, such as a list or string.

When to use: When we want to perform repeated actions to items in a list

```python
# Loop through a list of numbers and calculate the square of each number.
my_numbers = [1, 2, 3]

squared_numbers = []
for number in my_numbers: # goes through a list one item at a time
    # runs the code for each item.
    print('number:', number)
    square = number ** 2
    squared_numbers.append(square)
print('squared_numbers:', squared_numbers)
```

```python
# Be careful when loop through a list of mixed-type items
print('my_mixed_list:', my_mixed_list)

# for item in my_mixed_list:
#   print(item)
#   subtracted_item = item - 1 # This will break when the item is not int/float

# Prevent program from crashing
for item in my_mixed_list:
    if isinstance(item, int) or isinstance(item, float): # check whether data type is int or float
        subtracted_item = item - 1
    else:
        print('The item is not a int or a float. The item is', item)
```

```python
# range() returns an iterable (i.e., an object you can loop over)
for i in range(5):
    print(f'i=', i)
```

```python
# Use range() function for getting the index of each item
my_numbers = [1, 2, 100, 100, 200]
for i in range(len(my_numbers)):
    my_number_at_index_i = my_numbers[i]
    print('index =', i, 'my_number_at_index_i =', my_number_at_index_i)
```

```python
# enumerate() function
print('my_numbers:', my_numbers)
for i, item in enumerate(my_numbers):
    print('index =', i, 'item =', item)
```

# String: split(), strip(), lower(), upper(), title(), replace()

```python
# You can use split() to split a string into a list of sub-strings by given seperater
names = 'Esther|Andy|Emily'
names = names.split('|')
print('names:', names)

my_sentence = "What a lovely day!\nIt's windy outside."
words1 = my_sentence.split(' ')
print('words1:', words1)
words2 = my_sentence.split()
print('words2:', words2)
```

```python
punctuation
```

```python
# Use strip() to remove leading and trailing whitespace or specified characters.
# Wrap each string in a list so that whitespace is visible in the output.

my_name = ' Esther '
my_name_stripped = my_name.strip()
print('my_name:', [my_name])
print('my_name_stripped:', [my_name_stripped])

my_sentece = 'What a lovely day!'
my_sentece_sripped = my_sentece.strip(punctuation)
print('my_sentece:', my_sentece)
print('my_sentece_sripped:', my_sentece_sripped)

# lstrip()
my_name_lstripped = my_name.lstrip()
print('my_name:', [my_name])
print('my_name_lstripped:', [my_name_lstripped])

# rstrip()
my_name_rstripped = my_name.rstrip()
print('my_name:', [my_name])
print('my_name_rstripped:', [my_name_rstripped])
```

```python
#  Use replace() to replace occurrences of specified text with new text in a string.
my_name = 'Esther'
my_name_replaced = my_name.replace('E', 'CS')
print('my_name:', my_name)
print('my_name_replaced:', my_name_replaced)
```

```python
# lower(), upper(), title()
my_word = 'Hello'

my_word_lower = my_word.lower()
print('my_word:', my_word)
print('my_word_lower:', my_word_lower)

my_word_upper = my_word.upper()
print('my_word:', my_word)
print('my_word_upper:', my_word_upper)

my_word2 = 'python'
my_word2_title = my_word2.title()
print('my_word2:', my_word2)
print('my_word2_title:', my_word2_title)
```

## Read files
- read()
- readline()
- readlines()

```python
# This cell downloads an ebook and save it to a text file (moby_dick.txt).

url = "https://www.gutenberg.org/cache/epub/2701/pg2701.txt"

response = requests.get(url)
book = response.content.decode("utf-8-sig")

# 'w' in with open(...) means write mode. It opens a file for writing.
# If the file does not exist, Python creates it.
# If the file already exists, its contents are erased when it is opened.
with open("moby_dick.txt", "w", encoding="utf-8", newline="") as f:
    f.write(book)
```

```python
# Readline
# Read one line at a time. Each call reads the next line.

# 'r' in with open(...) means read mode. It opens a file for reading.
with open('moby_dick.txt', 'r') as f:
    first_line = f.readline()
    second_line = f.readline()

# Wrap each string in a list so that newline characters (\n) and whitespace are visible in the output.
print('first_line:', [first_line])
print('second_line:', [second_line])
```

```python
# Readlines
# Read all lines and return the lines as a list. Each item in the list is a line.

with open('moby_dick.txt', 'r') as f:
    data = f.readlines()

print('Number of lines in the file:', len(data))
print('The first two lines:', data[0:2])
```

```python
# Read
# Read all content in a file and return it as one string.

with open('moby_dick.txt', 'r') as f:
    data = f.read()

print('Number of characters in the file:', len(data))
print('The first 100 characters:', data[0:100])
```
