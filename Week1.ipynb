# Operations

## Operators
Operators are symbols, combinations of symbols, or keywords that designate some type of computation.

## Order of Operations
Python's order of operations is similar to what you might have learned in school, with a few tweaks.

- Parentheses have the highest precedence, so when in doubt put it in parens!
- Exponentiation has the next highest precedence.
- Multiplication, division, floor division, and the modulo have equal precendence
- Addition/subtraction have the lowest precendence of the aritmetic operators.

```python
# addition
my_number1 = 3
my_number2 = 4
sum_of_numbers = my_number1 + my_number2
print('sum_of_numbers:', sum_of_numbers)

# subtraction
subtracted_number = my_number1 - my_number2
print('subtracted_number:', subtracted_number)

# multiplication
multiplied_number = my_number1 * my_number2
print('multiplied_number:', multiplied_number)

# division
divided_number = my_number1 / my_number2
print('divided_number:', divided_number)

# exponentiation
exponentiated_number = my_number1 ** my_number2
print('exponentiated_number:', exponentiated_number)

# Note the order
my_number3 = my_number1 + my_number2 * my_number2
print('my_number3:', my_number3)
my_number4 = (my_number1 + my_number2) * my_number2
print('my_number4:', my_number4)

```

# Indexing
You can think of an index as a way to represent a character's position in a string.
Python indexes starting at zero

```python
# Example
my_word = 'Python'
print(my_word[0])
print(my_word[1])
print(my_word[-2])
```

# Slicing
Used for selecting a segment/range/sub-sequence of characters

```python
# Example
my_substring = my_word[0:2] # start: end
print('my_substring:', my_substring)
my_substring2 = my_word[ :2]
print('my_substring2:', my_substring2)
my_substring3 = my_word[2: ]
print('my_substring3:', my_substring3)
```

```python
# What will happen if I call my_string[:]?
my_word[ : ]
```

```python
# Slicing with step
print(my_word)
my_substring4 = my_word[0:6:2] # start: end: step
print('my_substring4:', my_substring4)
```

```python
# A slice extends beyond the length of a string
print(len(my_word))
print(my_word[1: 1000])
```

# Membership Operators & Boolean Expressions
The membership operator checks whether a value is present. The operators are `in` (and `not in`). They return a Boolean (`True` or `False`) value. You can also use Boolean expressions (`and`, `or`, and `not`) to add certain conditions to your statements.

```python
# Example -- in
print(my_word)
print('P' in my_word)
print('p' in my_word)
```

```python
# Example -- not in
print('p' not in my_word)
```

```python
# Example -- and
print('P' in my_word and 'y' in my_word)
print('p' in my_word and 'y' in my_word)
```

```python
# Example -- or
print('p' in my_word or 'y' in my_word)
```

```python
# Example -- not
print('p' not in my_word)
```

# if, elif, and else

`if`, `elif`, and `else` are keywords used in conditional statements. They control which block of code runs based on whether a condition is true or false.
- if: Checks the first condition.
- elif (short for “else if”): Checks another condition if the previous conditions were false.
- else: Runs if none of the conditions were true.



```python
# if/else example
my_number = 1

if my_number > 3:
    print('my_number is greater than 3')
else:
    print('my_number is not greater than 3')
```

```python
# if/elif/else example
my_number = 10

if my_number > 10:
    print('my_number is greater than 10')
elif my_number > 2:
    print('my_number is greated than 2')
elif my_number > 5:
    print('my_number is greater than 5')
else:
    print('my_number is less than or equal to 5')

# Note the logic: elif my_number > 5 is never checked
# because the earlier condition, elif my_number > 2, is already met.

# Here's the correct logic:
# if my_number > 10:
#     print('my_number is greater than 10')
# elif my_number > 5:
#     print('my_number is greater than 5')
# elif my_number > 2:
#     print('my_number is greated than 2')
# else:
#     print('my_number is less than or equal to 2')
```

```python
# if/else with membership operators
print(my_word)

if 'P' in my_word and 'y' in my_word:
    print('P and y are in my_word')
else:
    print('P and y are not in my_word')
```
